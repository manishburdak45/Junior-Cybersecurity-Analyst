## 1. Beginning of Networking

### Lecture Content

The lecture presents the beginning of computer networking as a sequence of events:

| Year | Event |
|---|---|
| 1959 | ARPA |
| 1960–1966 | Theoretical research |
| 1966 | ARPANET formally proposed |
| 1968 | BBN tasked to design interfacing processors |
| October 29, 1969 | First ARPANET connection between UCLA and SRI |

**Explanation of each point**

- **1959 — ARPA.** This is the starting point of the timeline.
- **1960–1966 — theoretical research.** Several years of research on the ideas of networking took place before any network was built. The work in this period was theoretical, not an operating network.
- **1966 — ARPANET formally proposed.** The research led to a formal proposal for ARPANET. This is the point at which ARPANET became a defined project.
- **1968 — BBN tasked to design interfacing processors.** BBN was given the task of designing the interfacing processors. The wording indicates that special processors were needed to interface the participating computers with the network.
- **October 29, 1969 — first ARPANET connection.** The first connection was established between **UCLA** and **SRI**. This is the first actual working connection of ARPANET.

**Significance of the sequence**

The order of events shows a clear development path: organization (ARPA), then theory, then a formal proposal, then design of the interfacing hardware, and finally a real connection. In other words, networking began as research and planning, and became a working system only after the proposal and design stages were complete.

**Chronological timeline**

```
1959
  |   ARPA
  v
1960–1966
  |   Theoretical research
  v
1966
  |   ARPANET formally proposed
  v
1968
  |   BBN tasked to design interfacing processors
  v
1969
      First ARPANET connection (UCLA and SRI), October 29
```

### Additional Understanding

The following points are general knowledge and are not part of the lecture content.

- **ARPA** stands for Advanced Research Projects Agency, a research agency of the United States Department of Defense. Many references give 1958 as the year ARPA was established. For answers that follow the lecture, use 1959 as given in the lecture.
- **BBN** stands for Bolt Beranek and Newman, a research and engineering company.
- The interfacing processors designed by BBN are commonly called **IMPs (Interface Message Processors)**.
- The theoretical research of the early 1960s is commonly associated with the development of **packet switching**, in which data is divided into small units that are sent through the network.

## 2. ARPANET

### Lecture Content

**What ARPANET was**

ARPANET was the network that was formally proposed in 1966 as an outcome of the theoretical research of 1960–1966. Its first connection was made on **October 29, 1969**, between **UCLA** and **SRI**.

**The first ARPANET connection**

| Item | Detail |
|---|---|
| Date | October 29, 1969 |
| Endpoints | UCLA and SRI |
| Significance | First ARPANET connection |

```
UCLA  ─────────────  SRI
```

The same connection drawn with a label:

```
UCLA
  |
  | Network Connection
  |
SRI
```

Two sites joined by one link is the simplest possible network. ARPANET began at this scale.

**Importance of ARPANET in the development of computer networking**

As shown by the lecture timeline, ARPANET is the central event of early networking history: the research of 1960–1966 led to the 1966 proposal, the 1968 design work by BBN, and the 1969 UCLA–SRI connection. The growth data in the next topic (4 nodes in 1969 to 23 nodes in 1972, and international connections in 1973) shows that this first connection was the start of a continuously expanding network.

**Note on the purpose of ARPANET**

The lecture points used for these notes record when and how ARPANET was proposed, designed, and first connected. They do not state a specific reason for its development. Any statement of purpose is therefore given below as Additional Understanding.

### Additional Understanding

The following points are general knowledge and are not part of the lecture content.

- **Why ARPANET was developed:** ARPANET was developed to allow research computers at different locations to be connected so that resources and information could be shared among them, and as an experiment in building a resilient network.
- **Names:** UCLA is the University of California, Los Angeles. SRI is the Stanford Research Institute.
- **First message:** The first message sent over the connection was an attempt to log in to the remote system. The system failed after the first two characters, "LO", and the full message was sent successfully later the same day.
- **Foundation of the Internet:** ARPANET is widely regarded as a direct predecessor of the Internet. Later in the 1970s and early 1980s, the protocols now known as TCP/IP were developed, and ARPANET moved to them in 1983. The ideas of connecting many different networks together grew out of the experience gained on ARPANET. This is general historical knowledge and is not stated in the lecture.

### Connection with NetSentinel-XDR

ARPANET is relevant to NetSentinel-XDR only at a conceptual level: it is an early example of interconnected computers communicating over a network, which is the environment that NetSentinel-XDR observes today. The full comparison, including what could and could not be verified, is given in the section titled **Connection with NetSentinel-XDR** later in this document.

## 3. Growth of ARPANET

### Lecture Content

The lecture gives the following growth data for ARPANET:

- The number of nodes increased each year from 1969 to 1972.
- In **1973**, the first **international connections** were made, to **Norway and the UK**.

**Year-by-year increase (calculated from the table)**

| Period | Increase in nodes |
|---|---|
| 1969 to 1970 | 9 (from 4 to 13) |
| 1970 to 1971 | 5 (from 13 to 18) |
| 1971 to 1972 | 5 (from 18 to 23) |

**What this growth demonstrates**

- **Rapid expansion.** The network grew from 4 nodes in 1969 to 23 nodes in 1972, which is more than five times as many nodes in three years.
- **Continuous growth.** The number of nodes increased in every year listed.
- **Geographic expansion.** The 1973 entry shows that the network extended beyond a single country, with first international connections to Norway and the UK.
- **A network can grow by adding nodes.** The four-node network of 1969 was extended to many nodes and, later, to other countries, rather than being replaced by a new design.

The 1973 row records an event, not a node count, and no node count is given for 1973 in the lecture. It must not be filled with a number.

### Additional Understanding

The following points are general knowledge and are not part of the lecture content.

- The international connection to Norway is commonly associated with NORSAR, and the connection to the UK with University College London.
- Growth in the number of hosts and networks continued well beyond 1973, and the interconnection of networks eventually led to what is now called the Internet.

## 4. Private Networks

### Lecture Content

After ARPANET, the lecture lists several **private networks** developed by companies and other organizations.

### Timeline/Table

| Year | Network | Organization/Company |
|------|---------|----------------------|
| 1974 | IBM SNA | IBM |
| 1975 | DECnet | DEC |
| 1975 | Telenet | — (not given in the lecture) |
| 1977 | Xerox XNS | Xerox |
| 1981 | BITNET | — (not given in the lecture) |

```
1974 ── IBM SNA
1975 ── DECnet
1975 ── Telenet
1977 ── Xerox XNS
1981 ── BITNET
```

**Reading the table**

- Five private networks are listed between 1974 and 1981.
- Three of them are associated in the lecture with a named company: IBM (SNA), DEC (DECnet), and Xerox (XNS).
- For Telenet and BITNET, the lecture does not name an organization, so none is stated here.
- Two networks, DECnet and Telenet, appear in the same year, 1975.

**Why the development of different private networking technologies was important**

- **Networking was no longer limited to ARPANET.** The list shows that, in the years after ARPANET, other organizations built their own networks.
- **Technology diversity.** Different networks had different names and came from different organizations, which means that several distinct networking technologies were in use at the same time.
- **Continued development of networking.** The spread of networks over the period 1974 to 1981 shows that networking was developing steadily after the first ARPANET connections.

### Additional Understanding

The following points are general knowledge and are not part of the lecture content.

- **SNA** is Systems Network Architecture. **DECnet** was developed by Digital Equipment Corporation. **XNS** is Xerox Network Systems.
- **Telenet** was a commercial packet-switched data network. **BITNET** was a cooperative network linking academic institutions, with early links in the United States.
- Many of these private networks were **proprietary**, meaning they were specific to one vendor or community and were generally not compatible with each other. This situation created the need to connect different networks together, which is the idea behind interconnected networks and, ultimately, the Internet.
